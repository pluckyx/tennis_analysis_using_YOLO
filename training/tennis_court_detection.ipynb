```python
import json
import os
import random
import cv2
import matplotlib.pyplot as plt

# 1. Load the JSON data
with open('data/data_train.json', 'r') as f:
    data = json.load(f)

# 2. Pick a random annotation
item = random.choice(data)
img_name = f"{item['id']}.png"
img_path = os.path.join('data/images', img_name)

# 3. Read the image
img = cv2.imread(img_path)
if img is None:
    print(f"Could not load image: {img_path} (Double check if the file exists)")
else:
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    
    plt.figure(figsize=(12, 8))
    plt.imshow(img)
    
    # 4. Loop through the keypoints and plot them
    kps = item['kps']
    for i, (x, y) in enumerate(kps):
        # Draw the point
        plt.scatter(x, y, color='red', s=50, edgecolors='white', zorder=2)
        # Draw the index number next to the point
        plt.text(x + 10, y + 10, str(i), color='yellow', fontsize=14, fontweight='bold', zorder=3)
    
    plt.title(f"Court Keypoints Visualization - {img_name}, shape: {img.shape}")
    plt.axis('off')
    plt.show()
```

```python
import torch
from torchvision import transforms, datasets, models
from torch import nn
from torch.utils.data import DataLoader
import json
from torchvision.transforms import v2
import cv2
import numpy as np
```

```python
device = "cuda" if torch.cuda.is_available() else 'cpu'
device
```

```python
print(torch.cuda.get_device_name(0))
```

```python
class KeyPoints:
    def __init__(self, img_dir, data_file):
        self.img_dir = img_dir
        with open (data_file, "r") as f:
            self.data = json.load(f)
        self.transforms = v2.Compose([
            v2.ToImage(),
            v2.Resize((224,224)),
            v2.ToDtype(torch.float32,scale=True),
            v2.Normalize(mean=[0.485, 0.456, 0.406],std=[0.229, 0.224, 0.225])
        ])
    
    def __len__(self):
        return len(self.data)
    
    def __getitem__(self, index):
        item = self.data[index]
        img = cv2.imread(f"{self.img_dir}/{item['id']}.png")
        H, W  = img.shape[:2]
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        img = self.transforms(img)
        kps = np.array(item['kps']).flatten()
        kps = kps.astype(np.float32)
        kps[::2] = kps[::2] * 224.0 / W
        kps[1::2] = kps[1::2] * 224.0 / H


        return img, kps
```

```python
train_dataset = KeyPoints("data/images","data/data_train.json")
train_dataset 
valid_dataset = KeyPoints("data/images","data/data_val.json")
valid_dataset
```

```python
train_loader = DataLoader(train_dataset,batch_size=8,shuffle=True)
valid_loader = DataLoader(valid_dataset,batch_size=8,shuffle=True)
```

```python
# Install torchinfo if it's not available, import it if it is
try: 
    import torchinfo
except:
    !pip install torchinfo
    import torchinfo
    
from torchinfo import summary
```

```python
weights = models.ResNet50_Weights.DEFAULT
model = models.resnet50(weights=weights)
```

```python
# Print a summary using torchinfo (uncomment for actual output)
summary(model=model, 
        input_size=(8, 3, 224, 224), # make sure this is "input_size", not "input_shape"
        # col_names=["input_size"], # uncomment for smaller output
        col_names=["input_size", "output_size", "num_params", "trainable"],
        col_width=20,
        row_settings=["var_names"]
) 
```

Linear layer called Fc

```python
for param in model.parameters():
    param.requires_grad = True
```

```python
model.fc = torch.nn.Linear(in_features=model.fc.in_features, out_features= 14*2) #how many points on x,y
```

```python
summary(model=model, 
        input_size=(8, 3, 224, 224), # make sure this is "input_size", not "input_shape"
        # col_names=["input_size"], # uncomment for smaller output
        col_names=["input_size", "output_size", "num_params", "trainable"],
        col_width=20,
        row_settings=["var_names"])
```

```python
criterion = torch.nn.MSELoss()
```

```python
optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)
```

```python
from tqdm.auto import tqdm
```

```python
model = model.to(device)
best_val_loss = float('inf')
save_path = "best_model.pt"

for epoch in tqdm(range(10)):
    # ---- TRAINING ----
    model.train()
    train_loss = 0.0
    for i, (img, kps) in tqdm(enumerate(train_loader)):
        img = img.to(device)
        kps = kps.to(device)

        optimizer.zero_grad()
        output = model(img)
        loss = criterion(output, kps)
        loss.backward()
        optimizer.step()

        train_loss += loss.item()

        if i % 100 == 0:
            print(f"  Epoch {epoch+1} | iter {i}/{len(train_loader)} | loss: {loss.item():.4f}")

    avg_train_loss = train_loss / len(train_loader)

    # ---- VALIDATION ----
    model.eval()
    val_loss = 0.0
    total_kps = 0
    accurate_kps = 0
    threshold = 5.0

    with torch.no_grad():
        for img, kps in valid_loader:
            img = img.to(device)
            kps = kps.to(device)

            output = model(img)
            loss = criterion(output, kps)
            val_loss += loss.item()

            pred = output.view(-1, 14, 2)
            target = kps.view(-1, 14, 2)
            distances = torch.sqrt(((pred - target) ** 2).sum(dim=2))
            accurate_kps += (distances < threshold).sum().item()
            total_kps += distances.numel()

    avg_val_loss = val_loss / len(valid_loader)
    accuracy = 100.0 * accurate_kps / total_kps

    print(f"Epoch {epoch+1} DONE | Train Loss: {avg_train_loss:.4f} | Val Loss: {avg_val_loss:.4f} | Acc (<{threshold}px): {accuracy:.2f}%")

    if avg_val_loss < best_val_loss:
        best_val_loss = avg_val_loss
        torch.save(model.state_dict(), save_path)
        print(f"Best model saved! (val_loss: {best_val_loss:.4f})")

print(f"\nDone! Best val loss: {best_val_loss:.4f}")
```
